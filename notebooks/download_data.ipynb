### 4a. Dataset 1 — Fake News (True vs Fake Articles)

```python
# ~41 MB — downloads to data/fake-and-real-news-dataset.zip
# Renamed to: data/fake _news.zip
import subprocess
result = subprocess.run(
    ['kaggle', 'datasets', 'download',
     '-d', 'clmentbisaillon/fake-and-real-news-dataset',
     '-p', 'data/', '--force'],
    capture_output=True, text=True
)
print(result.stdout)
if result.returncode == 0:
    # Rename to expected name
    src = 'data/fake-and-real-news-dataset.zip'
    dst = 'data/fake _news.zip'
    if os.path.exists(src) and not os.path.exists(dst):
        os.rename(src, dst)
    print('Dataset 1 ready:', os.path.getsize('data/fake _news.zip') / 1e6, 'MB')
else:
    print('ERROR:', result.stderr)
```

## Step 6: Full Verification — All 4 Datasets

```python
import os, zipfile
import pandas as pd

print('=' * 70)
print('FULL DATASET VERIFICATION — ALL 4 DATASETS')
print('=' * 70)

results = []

# ── Dataset 1: Fake News ─────────────────────────────────────────────
try:
    with zipfile.ZipFile('data/fake _news.zip', 'r') as z:
        with z.open('Fake.csv') as f: n_fake = len(pd.read_csv(f))
        with z.open('True.csv') as f: n_real = len(pd.read_csv(f))
    ok = (n_fake == 23481 and n_real == 21417)
    results.append(('Fake News', ok, f'Fake:{n_fake:,} Real:{n_real:,} Total:{n_fake+n_real:,}'))
except Exception as e:
    results.append(('Fake News', False, str(e)))

# ── Dataset 2: Phishing Emails ───────────────────────────────────────
try:
    with zipfile.ZipFile('data/phishing_emails.zip', 'r') as z:
        with z.open('phishing_email.csv') as f: n_p = len(pd.read_csv(f))
    ok = (n_p == 82486)
    results.append(('Phishing Emails', ok, f'Total: {n_p:,} emails'))
except Exception as e:
    results.append(('Phishing Emails', False, str(e)))

# ── Dataset 3: UNSW-NB15 ─────────────────────────────────────────────
try:
    with zipfile.ZipFile('data/behavioral_anomalies.zip', 'r') as z:
        with z.open('UNSW_NB15_training-set.csv') as f: n_tr = len(pd.read_csv(f))
        with z.open('UNSW_NB15_testing-set.csv') as f:  n_te = len(pd.read_csv(f))
    ok = (n_tr == 82332 and n_te == 175341)
    results.append(('UNSW-NB15', ok, f'Train:{n_tr:,} Test:{n_te:,} Total:{n_tr+n_te:,}'))
except Exception as e:
    results.append(('UNSW-NB15', False, str(e)))

# ── Dataset 4: Deepfake ───────────────────────────────────────────────
try:
    counts = {'train':0,'valid':0,'test':0}
    with zipfile.ZipFile('deepfake.zip', 'r') as z:
        for n in z.namelist():
            p = n.split('/')
            if len(p) >= 4 and p[2] in counts and p[3] in ('fake','real'):
                counts[p[2]] += 1
    total = sum(counts.values())
    ok = (total == 140000)
    results.append(('Deepfake Faces', ok,
                    f"Train:{counts['train']:,} Valid:{counts['valid']:,} "
                    f"Test:{counts['test']:,} Total:{total:,}"))
except Exception as e:
    results.append(('Deepfake Faces', False, str(e)))

# ── Print results ─────────────────────────────────────────────────────
print()
all_ok = True
for name, ok, detail in results:
    status = 'PASS' if ok else 'FAIL'
    all_ok = all_ok and ok
    print(f'  [{status}] {name}')
    print(f'         {detail}')
    print()

print('=' * 70)
if all_ok:
    print('ALL 4 DATASETS VERIFIED SUCCESSFULLY')
else:
    print('SOME DATASETS FAILED — re-download missing files')
print('=' * 70)
```
